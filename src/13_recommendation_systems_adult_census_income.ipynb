```python
df = df.drop(columns=["fnlwgt", "capital.gain", "capital.loss", "marital.status", "relationship", "education"])
df.head(5)
```

#### Cambiar valores

```python
df.replace('?', np.nan, inplace=True)
```

```python
df['income']
```

```python
df['income'] = df['income'].str.strip()
df['income'] = df['income'].apply(lambda x: 1 if x == '>50K' else 0)
```

```python
df['income']
```

```python
# realizando one-hot-encoding para pasar a numerico las variables de clasificacion
categorical_cols = ["workclass","occupation", "sex", "race", "native.country"]
df_encoded = pd.get_dummies(df, columns=categorical_cols, dtype=int)
df_encoded
```

```python
# Normalizacion de las variables numericas

scaler = StandardScaler()

df_encoded[['age', 'hours.per.week','education.num']] = scaler.fit_transform(
    df_encoded[['age', 'hours.per.week', 'education.num']])
df_encoded.head()
```

#### split

```python
X = df_encoded.drop(columns=['income'])
y = df_encoded['income']
```

```python
# calcular similitudes entre usuarios
similarity_matrix = cosine_similarity(X)
```

```python


# Funcion de recomendacion
# Busca personas similares que ganan mas de 50k y recomienda sus caracteristicas

def recomendar_trayectoria(usuario_idx, df_original, X, y, top_k=5):
    sims = similarity_matrix[usuario_idx]

    similar_users = sims.argsort()[::-1][1:]
    similares_altos_ingresos = [
        i for i in similar_users if y.iloc[i] == 1
    ][:top_k]

    return df_original.iloc[similares_altos_ingresos][
        ['education.num', 'occupation', 'hours.per.week']]

```

```python
# Prueba simple: usar un usuario real del dataset
# Paso 1: elegir un usuario con ingreso bajo del mismo dataset

usuario_test = df[df['income'] == 0].index[0]
```

```python

df.loc[usuario_test]
```

```python
recomendar_trayectoria(usuario_test, df, X, y)
```

El sistema sugiere que para perfiles similares, el éxito financiero está asociado con un aumento en la carga horaria hacia las 40 horas semanales en sectores técnicos u operativos.

Al implementar la similitud de coseno, el modelo no solo encuentra personas parecidas, sino que actúa como una herramienta de consultoría de carrera. Al filtrar únicamente a los perfiles con ingresos altos (income == 1), el sistema es capaz de sugerir cambios específicos en la trayectoria de un usuario.

El modelo encontró personas que se parecen mucho a la mujer de 82 años (misma educación o perfil), pero que sí ganan bien.

Para este perfil específico, el modelo descubrió que quienes ganan más de 50k suelen trabajar 40 horas a la semana (tiempo completo) en lugar de 18, y se dedican a labores técnicas como inspección o reparación.

La educación y el puesto importan: El análisis confirma que lo que estudiaste (education.num) y en qué trabajas (occupation) son las llaves principales para predecir si alguien tendrá ingresos altos.
